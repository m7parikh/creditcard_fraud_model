```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
train_data = pd.read_csv("data/fraudTrain.csv")
test_data = pd.read_csv("data/fraudTest.csv")

print(f"Train data shape: {train_data.shape}")
print(f"Test data shape: {test_data.shape}")
train_data.head()
```

```python
train_data.info()
```

```python
train_data.describe()
```

```python
train_counts = train_data["is_fraud"].value_counts()
test_counts = test_data["is_fraud"].value_counts()

train_percent = train_data["is_fraud"].mean() * 100
test_percent = test_data["is_fraud"].mean() * 100

print(f"Train: {train_counts.to_dict()}, {train_percent:.2f}% fraud")
print(f"Test: {test_counts.to_dict()}, {test_percent:.2f}% fraud")
```

```python
#Check missing values
train_data.isna().sum()[train_data.isna().sum() > 0]
```

```python
#See how transaction amount varies between fraud and non-fraud cases
train_data.groupby("is_fraud")["amt"].describe()
```

```python
fig, ax = plt.subplots(figsize=(8,4))
sns.boxplot(data=train_data, x="is_fraud", y="amt", showfliers=False, ax=ax)
ax.set_yscale("log")
ax.set_title("Transaction amount by fraud label (log scale, outliers hidden)")
plt.show()
```

```python
category_fraud = (
    train_data.groupby("category")["is_fraud"]
    .agg(["mean", "count"])
    .rename(columns={"mean": "fraud_rate", "count": "n"})
    .sort_values("fraud_rate", ascending=False)
)

category_fraud["fraud_rate"] = (category_fraud["fraud_rate"] * 100).round(3)
category_fraud = category_fraud.reset_index()
category_fraud
```
